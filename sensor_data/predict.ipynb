```python
from os import name

import joblib

model = joblib.load('/content/drive/MyDrive/sensor_data/pet_health_model.pkl')

print(" PET HEALTH PREDICTOR")

name = input("Enter Name: ")
breed = input("Enter Breed: ")
age = float(input("Age (Years): "))
weight = float(input("Weight (kg): "))
temp = float(input("Temperature (°C): "))
heartbeat = int(input("Heartbeat (BPM): "))

if 'cat' in breed.lower() or 'siamese' in breed.lower() or 'persian' in breed.lower():
    breed_code = 1
else:
    breed_code = 0

# Prediction
input_data = [[breed_code, age, weight, temp, heartbeat]]
result = model.predict(input_data)[0]

print(f"\n Health Status: {result}")

joblib.dump(model, 'pet_health_model.pkl')
print(f"\n Model saved as 'pet_health_model.pkl'")
```
